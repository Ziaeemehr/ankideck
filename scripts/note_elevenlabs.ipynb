```python
from elevenlabs.client import ElevenLabs
from elevenlabs.play import play
import os


def audio_from_elevenlabs(text, api_key_address, filename, path="."):
    """
    Generate audio from text using ElevenLabs API.
    store audio as MP3 file.
    """
    # Load API key
    with open(api_key_address, "r", encoding="utf-8") as f:
        elevenlabs_api_key = f.read().strip()

    # Initialize ElevenLabs client
    elevenlabs = ElevenLabs(api_key=elevenlabs_api_key)

    # Generate audio
    audio = elevenlabs.text_to_speech.convert(
        text=text,
        voice_id="JBFqnCBsd6RMkjVDRZzb",
        model_id="eleven_multilingual_v2",
        output_format="mp3_44100_128",
    )
    # Save audio to MP3 file
    audio_bytes = b"".join(audio)
    os.makedirs(path, exist_ok=True)
    path_filename = f"{path}/{filename}"
    with open(path_filename, "wb") as f:
        f.write(audio_bytes)
    
    # return audio_bytes


# --- Load API Key directly from file ---
API_KEY_FILE = "/Users/tng/Projects/Language/FR/Anki_decks/elevenlabs_api_key.txt"

audio_from_elevenlabs(
    text="Ceci est un autre exemple.",
    api_key_address=API_KEY_FILE,
    filename="another_example.mp3",
    path="outputs",
)
```
